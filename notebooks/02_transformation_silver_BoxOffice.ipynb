##Importing data from Bronze Table

```
df_bronze = spark.read.format('delta').load('/FileStore/tables/Marvel/Bronze/BoxOffice_Performance/')
```

```
display(df_bronze)
```

##Excluding null columns

```
df_silver = df_bronze.select('movie_id', 'movie_name', 'Phase_id', 'worldwide_box_office', 'domestic_box_office', 'international_box_office', 'opening_weekend', 'production_budget', 'worldwide_box_office_millions', 'domestic_box_office_millions', 'international_box_office_millions', 'opening_weekend_millions','production_budget_millions', 'Budget_Percentage', 'Opening_Weekend_Gross_Percentage', 'Domestic_Gross_Percentage', 'International_Gross_Percentage','Budget_Opening_Weekend_Percentage', 'Budget_Recovered_Percentage', 'International_Collection_Percentage', 'International_vs_Budget_Percentage','Opening_Weekend_vs_Domestic_Percentage', 'Domestic_vs_Budget_Percentage', 'Post_Opening_Weekend_Percentage', 'Domestic_Post_Opening_Percentage','Post_Opening_vs_Budget_Percentage')
```

##Writing Delta Format Silver Layer

```
df_silver.write.format("delta").mode("overwrite").save("/FileStore/tables/Marvel/Silver/BoxOffice_Performance/")
```
