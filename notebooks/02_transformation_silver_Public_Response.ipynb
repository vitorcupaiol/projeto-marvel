##Importing functions

```
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```
dbutils.fs.ls('/FileStore/tables/Marvel/Bronze/')
```

##Importing data from Bronze Table

```
df_bronze = spark.read.format('delta').load('/FileStore/tables/Marvel/Bronze/public_response/')
```

```
display(df_bronze)
```

######Changing 'NULL' values to zero on then column Audience_Score and converting to integer

```
df_silver = df_bronze.withColumn('audience_score',when(col('audience_score')=='NULL',0).otherwise(col('audience_score')).cast('integer'))
```

##Writing Delta Format Silver Layer

```
df_silver.write.format("delta").mode("overwrite").save("/FileStore/tables/Marvel/Silver/Public_Response/")
```
