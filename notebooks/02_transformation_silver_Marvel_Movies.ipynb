##Importing functions

```
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

##Importing data from Bronze Table

```
df_bronze = spark.read.format('delta').load('/FileStore/tables/Marvel/Bronze/marvel_movies/')
```

##Renaming Columns

```
df_bronze = df_bronze.withColumnsRenamed({'movie_id':'id','movie_name':'name'})
df_silver = df_bronze
```

##Writing Delta Format Silver Layer

```
df_silver.write.format("delta").mode("overwrite").save("/FileStore/tables/Marvel/Silver/Marvel_Movies/")
```
