##Importing data from Bronze Table

```
df_bronze = spark.read.format('delta').load('/FileStore/tables/Marvel/Bronze/Phases/')
```

##Renaming columns

```
df_bronze = df_bronze.withColumnRenamed('phase_id','id')\
  .withColumnRenamed('phase_name','name')
df_silver = df_bronze
```

##Writing Delta Format Silver Layer

```
df_silver.write.format("delta").mode("overwrite").save("/FileStore/tables/Marvel/Silver/Phases/")
```
